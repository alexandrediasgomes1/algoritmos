```python
# Nível 1 - Exercício 1 - Criando e acessando listas

lista = [7.3, 8.2, 4.2, 6.7, 4, 6.3]
print(f'A primeira nota é {lista[0]}')
print(f'A última nota é {lista[-1]}')
print(f'O total de notas é de {len(lista)}')
```

```python
# Nível 1 - Exercício 2 — Média simples

lista = [3.5, 6.7, 8.8, 9, 4.5]
media = sum(lista) / len(lista)
print('A média é', media)
```

```python
# Nível 2 - Exercício 4 - Vendas da semana
vendas = [1200.50, 980.00, 1350.75, 1100.00, 1580.90]
media = sum(vendas) / len(vendas)
maior = max(vendas)
menor = min(vendas)
print(f'O total vendido foi de R$ {sum(vendas)}')
print(f'A média de vendas foi de R$ {media:.2f}')
print(f'A maior venda foi de R$ {maior:.2f}')
print(f'A menor venda foi de R$ {menor:.2f}')



```
