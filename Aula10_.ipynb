```python
# Nível 4 - Exercício 10 - mini avaliador de frete
# esse deu trabalho!

compra = float(input('Digite o valor da compra'))

if compra >= 300:
  print('O frete é gratuito')
else:
  regiao = int(input('Digite a região de entrega: [1] para local, [2] para estadual e [3] para nacional'))
  if regiao == 1:
      print('O frete é de R$ 10,00')
      regiao = 10
  elif regiao == 2:
        print('O frete é de R$ 25,00.')
        regiao = 25
  elif regiao == 3:
        print('O frete é de R$ 50,00')
        regiao = 50
  total = compra + regiao

print('o valor total da compra é de R$ ', total)


```
