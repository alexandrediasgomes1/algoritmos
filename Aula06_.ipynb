```python
# Exercício 4 - Classificador de temperaturas
temp = float(input('Digite a temperatura'))
if temp < 15:
  print('Muito frio')
elif temp >= 15 and temp <= 25:
  print('Temperatura normal')
else:
  print('Muito quente')
```

```python
# Exercício_05 - somar de 1 até 10.
cont = 0
soma = 0
while cont <= 10:
    soma = soma + cont
    cont = cont + 1
print(soma)

```

```python
# Exercício_06 - tabuada de um número.

multiplo = 0
numero = int(input('Digite um número para ver a sua tabuada'))
while multiplo <= 10:
     tabuada = numero * multiplo
     print(f'{numero} x {multiplo} = {tabuada}')
     multiplo = multiplo + 1
```

```python
# Exercício 9 - Fatorial de um número.
num = int(input('Digite um número para calcular o seu fatorial'))
cont = 1
fat = 1
while cont <= num:
  fat = fat * cont
  cont = cont + 1
print(fat)

```
