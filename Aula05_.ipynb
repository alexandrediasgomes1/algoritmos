```python
# Exercício_01 - Verificar se é de maior.
idade = int(input('Digite a sua idade'))
if idade >= 18:
  print('Você é maior de idade')
else:
  print('Você é menor de idade')
```

```python
# Exercício_02 - Somar os itens da compra.
p1 = int(input('Digite o valor do 1º produto '))
p2 = int(input('Digite o valor do 2º produto '))
p3 = int(input('Digite o valor do 3º produto '))
soma = p1 + p2 + p3
print('O valor total da compra é de',soma)
```

```python
# Exercício_05 - Verificar se número = positivo, negativo ou zero.
n = int(input('Digite um número'))
if n < 0:
  print('Número negativo')
elif n > 0:
    print('Número positivo')
else:
      print('Número zero')


```

```python

# Exercício_06 - valor de desconto.
valor = int(input('digite o valor da compra'))
if valor > 200:
  valor = valor - valor * 0.10
  print('o novo valor é de ', valor)
else:
  print('o valor é de ', valor)

```

```python
# Exercício_10 - calcular IMC.
peso = float(input('Digite o seu peso'))
altura = float(input('Digite a sua altura'))
imc = peso / altura**2
if imc < 18.5:
  print('O seu IMC é de', imc)
  print('Abaixo do peso')
elif imc >= 18.5 and imc <= 24.9:
  print('O seu IMC é de', imc)
  print('Peso normal')
else:
  print('O seu IMC é de', imc)
  print('Acima do peso')
```
