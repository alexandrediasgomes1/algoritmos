```python
# Nível 4 - Exercício 10 - Processamento de turmas com nomes

aprovado = 0
reprovado = 0
recuperacao = 0
soma_medias = 0
media_geral = 0
maior_media = -1.0  # Iniciamos com um valor bem baixo para que qualquer média seja maior
menor_media = 11.0  # Iniciamos com um valor bem alto para que qualquer média seja menor

alunos = int(input("Quantos alunos tem na turma? "))
if alunos <= 0:
    print("Quantidade de alunos inválida")
else:
  for c in range(alunos):
    nome = input("Digite o nome do aluno: ")
    nota1 = float(input("Digite a nota 1: "))
    nota2 = float(input("Digite a nota 2: "))
    nota3 = float(input("Digite a nota 3: "))
    nota4 = float(input("Digite a nota 4 : "))

    media_ponderada = (nota1 * 0.3) + (nota2 * 0.3) + (nota3 * 0.2) + (nota4 * 0.2)
    print("A média do aluno", nome, "é", media_ponderada)
    soma_medias += media_ponderada
    if media_ponderada > maior_media:
        maior_media = media_ponderada
    if media_ponderada < menor_media:
        menor_media = media_ponderada
    if media_ponderada >= 7:
        aprovado = aprovado + 1
    elif media_ponderada >= 5:
        recuperacao = recuperacao + 1
    else:
        reprovado = reprovado + 1

  print('Resumo da turma')
  print('Quantidade de alunos aprovados:', aprovado)
  print('Quantidade de alunos em recuperação:', recuperacao)
  print('Quantidade de alunos reprovados:', reprovado)

  media_geral = soma_medias / alunos
  print('Média geral da turma:', media_geral)
  print('A maior média da turma é:', maior_media)
  print('A menor média da turma é:', menor_media)
```
