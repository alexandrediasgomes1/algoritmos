```python
# Exercício 5 - Ticket médio
venda_total = float(input('Digite o valor total das vendas'))
tickets = float(input('Digite a quantidade de tickets vendidos'))
ticket_medio = venda_total / tickets
print(f'O valor médio das vendas é de {ticket_medio:.2f}')
```

```python
# Exercício 6 - Conversão de temperatura
# F = C × 9 / 5 + 32
cel = float(input('Digite o valor da temperatura'))
fah = cel * 9 / 5 + 32
print(f'A temperatura de {cel} graus Celsius convertida para Fahrenhei fica em {fah:.1f} graus ')
```

```python
# Exercício 7 - Média de 3 notas
nota1 = float(input('Digite a primeira nota'))
nota2 = float(input('Digite a segunda nota'))
nota3 = float(input('Digite a terceira nota'))
media = (nota1 + nota2 + nota3) / 3
print(f'O valor da média foi de {media:.2f}')
```

```python
# Exercício 12 - Desconto do produto
produto = str(input('Digite o nome do produto'))
preco = float(input('Digite o preço do produto'))
desconto = float(input('Digite o desconto do produto %'))
desconto = preco * desconto / 100
novo_preco = preco - desconto
print(f'O produto {produto} teve um desconto de {desconto} é o seu novo preço é de R${novo_preco:.2f}')
```
