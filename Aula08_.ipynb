```python
# Nível 1 - Exercício 1 - 4 operações com 2 números
n1 = (int(input("Digite o primeiro número: ")))
n2 = (int(input("Digite o segundo número: ")))
soma = n1 + n2
subtracao = n1 - n2
multiplicacao = n1 * n2
divisao = n1 / n2
print(f'A soma dos números é: {soma}')
print(f'A subtração dos números é: {subtracao}')
print(f'A multiplicação dos números é: {multiplicacao}')
print(f'A divisão dos números é: {divisao:.2f}')
```

```python
# Nível 2 - Exercício 4 - Preço e porcentagem
preco = float(input("Digite o preço do produto: "))
porcentagem = float(input("Digite a porcentagem de desconto: "))
if preco > 0 and porcentagem > 0 and porcentagem <= 50:
    desconto = preco * (porcentagem / 100)
    preco_com_desconto = preco - desconto
    print(f"O preço com desconto é: R${preco_com_desconto:.2f}")
else:
  print('Esse produto não tem desconto.')

```

```python
# Nível 4 - Exercício 10 - Formar e classificar um triângulo
l1 = float(input("Digite o primeiro lado do triângulo: "))
l2 = float(input("Digite o segundo lado do triângulo: "))
l3 = float(input("Digite o terceiro lado do triângulo: "))
if l1 < l2 + l3 and l2 < l1 + l3 and l3 < l1 + l2:
  print('É possível formar um triângulo.')
if l1 == l2 == l3:
  print('É um triângulo equilátero.')
elif l1 == l2 or l1 == l3 or l2 == l3:
  print('É um triângulo isósceles.')
else:
  print('É um triângulo escaleno')
```

```python
# Nível 5 IA sob auditoria - Exercício 13 - Conversor de temperatura
celsius = float(input("Digite a temperatura em Celsius: "))

fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15

print(f"{celsius:.2f}°C equivalem a:")
print(f"{fahrenheit:.2f}°F")
print(f"{kelvin:.2f}K")
```
