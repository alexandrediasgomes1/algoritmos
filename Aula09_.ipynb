```python
# Nível 2 - Exercício 4 - Desconto em compras
print('GANHE UM DESCONTO DE 10% PARA A COMPRAS ACIMA DE R$ 200,00 REAIS')
preco = int(input('Digite o valor da compra:'))
desc = preco * 0.10
if preco >= 200:
  preco = preco - desc
  print(f'O novo preço com o desconto será de {preco:.2f}')
else:
  print('Você não tem direito ao desconto.')
  print('Valor da compra:',preco)


```

```python
# Nível 4 - Exercício 10 - Média Ponderada das notas

p1 = float(input('Digite a primeira nota:'))
p2 = float(input('Digite a segunda nota:'))
p3 = float(input('Digite a terceira nota:'))
p4 = float(input('Digite a quarta nota:'))
media = (p1 * 0.3) + (p2 * 0.3) + (p3 * 0.2) + (p4 * 0.2) / 4
if media >= 7:
  print('Aluno Aprovado')
elif media <= 5:
  print('Aluno em recuperação')
  nr = float(input('Digite a nota de recuperação:'))
  media = (media + nr) / 2
  if media >= 5:
    print('Aluno Aprovado')
  else:
    print('Aluno Reprovado!')
else:
  print('Aluno Reprovado!')
```
